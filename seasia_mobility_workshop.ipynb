### ① 人口データを決める

|---|---|
| H1〜H4の人口（町全体で100人） | 施設の位置、道路網 |
| 住宅群ごとの活動タイプの割合（それぞれ合計100％） | 活動タイプごとの外出予定・訪問先の選び方 |
| 活動タイプ：学生／通勤者／農業従事者／その他 | 徒歩の速さ、昼休み、買い物、個人差の扱い |

**基準人口：** H1〜H4に各25人。町全体では学生20・通勤者30・農業従事者30・その他20人です。

住宅群の人口を変えると、その住宅群の活動タイプ別人数も、設定した割合に応じて変わります。町全体の活動タイプ別人数は固定されません。

**人の動きを決める条件**  人口データのほかに、以下の条件も必要です。この演習では、すべてあらかじめ決めてあります。

| 項目 | あらかじめ決めてある内容 |
|---|---|
| 主な行き先 | 学生は学校、通勤者は事業所、農業従事者は農地へ。「その他」は主に自宅で過ごす。訪問先は近いほど選ばれやすいが、遠くを選ぶ人もいる |
| 自宅を出る／主な活動場所を出る時刻 | 学生 **7:00／15:30**、通勤者 **7:30／17:00**、農業従事者 **6:00／16:30** が基本。人によって最大10分ずれる |
| 昼休み | 基準人口では農業従事者30人中9人が12:00〜14:00に一度帰宅し、ほかは農地で休む。人口設定変更後は帰宅者が全農業従事者の30％とは限らない |
| 買い物 | 通勤者・農業従事者の約4割が帰りに店へ寄り30分滞在。「その他」の約7割は8:30〜16:30の間に買い物へ出て45分滞在 |
